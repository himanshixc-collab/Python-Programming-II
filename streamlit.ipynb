CALCULATOR

```python
%pip install streamlit
import streamlit as st

st.set_page_config(page_title="Calculator App", page_icon="🤓☝️")

st.title("🧮 Calculator App")
st.write("Simple Calculator deployed on Streamlit")

num1 = st.number_input("Enter first number", value=0.0)
num2 = st.number_input("Enter second number", value=0.0)

operation = st.radio(
    "Select Operation",
    ("Addition", "Subtraction", "Multiplication", "Division")
)

if st.button("Calculate"):
    if operation == "Addition":
        st.success(f"Result: {num1 + num2}")

    elif operation == "Subtraction":
        st.success(f"Result: {num1 - num2}")

    elif operation == "Multiplication":
        st.success(f"Result: {num1 * num2}")

    elif operation == "Division":
        if num2 == 0:
            st.error("❌ Division by zero is not possible")
        else:
            st.success(f"Result: {num1 / num2}")
```

FORM

```python
import streamlit as st

st.title("Student Information Form")

# Text input for name
# User enters a single line of text
name = st.text_input("Enter your name")

# Number input for age
# User enters only numbers
age = st.number_input("Enter your age")

# Dropdown selection
# User selects one option
course = st.selectbox(
    "Select the Course you are currently studying",
    ["BCA","BBA","MBA","MCA"]
)

# Radio button
# Select only one option
gender = st.radio(
    "Select Gender",
    ["Male","Female","Other"]
)


#multiselect
skills = st.multiselect(
    "Select your skills",
    ["Python", "Java", "C++", "AI", "Machine Learning","Other"]
)
# Slider input
# User drags slider between values
marks = st.slider(
    "Enter total percentage",
    0,100
)

# Checkbox
agree = st.checkbox(
    "I agree to terms and conditions"
)


# Button
if st.button("Submit"):

    st.write("Name:",name)
    st.write("Age:",age)
    st.write("Course:",course)
    st.write("Gender:",gender)
    st.write("Total percentage:",marks)
    st.write("skills:",skills)

    if agree:
        st.success("Form Submitted")
    else:
        st.error("Please accept terms"
)
```

BMI CALCULATOR

```python
import streamlit as st

st.title("BMI Calculator")

weight = st.number_input("Weight (kg)")
height = st.number_input("Height (m)")

if st.button("Calculate BMI"):
    bmi = weight/(height**2)
    st.success(f"BMI: {bmi:.2f}")

    if bmi<18.5:
        st.warning("Underweight")
    elif bmi<25:
        st.success("Normal")
    else:
        st.error("Overweight")
```

AGE CALCULATOR

```python
import streamlit as st
from datetime import date

st.title("Age Calculator")

dob = st.date_input("Select DOB")

if st.button("Calculate"):
    today=date.today()
    age=today.year-dob.year

    st.success(f"Age: {age} years")
```

PASSWORD STRENGTH CHECKER

```python
import streamlit as st

st.title("Password Checker")

pwd=st.text_input(
    "Password",
    type="password"
)

if st.button("Check"):

    score=0

    if len(pwd)>=8:
        score+=1

    if any(c.isdigit() for c in pwd):
        score+=1

    if any(c.isupper() for c in pwd):
        score+=1

    st.write("Strength:",score,"/3")
```
