```python
!wget -q https://dl.google.com/linux/direct/google-chrome-stable_current_amd64.deb
!apt -qq install -y ./google-chrome-stable_current_amd64.deb

!pip install -q selenium webdriver-manager

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

from selenium.webdriver.chrome.options import Options
import time
import pandas as pd

chrome_options = Options()
chrome_options.add_argument("--headless=new")
chrome_options.add_argument("--no-sandbox")
chrome_options.add_argument("--disable-dev-shm-usage")

browser = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=chrome_options
)

browser.get("https://www.bing.com/")
time.sleep(2)

search_box = browser.find_element(By.NAME, "q")
search_box.send_keys("AAFT")
search_box.send_keys(Keys.ENTER)

time.sleep(4)

print("Page Loaded Successfully\n")
print(browser.title)

search_results = browser.find_elements(By.CSS_SELECTOR, "li.b_algo")

records = []

for item in search_results[:5]:
    try:
        heading = item.find_element(By.TAG_NAME, "h2").text
        url = item.find_element(By.TAG_NAME, "a").get_attribute("href")
        records.append([heading, url])
    except:
        pass

print("\nTop Bing Search Results:\n")

for count, row in enumerate(records, start=1):
    print(f"{count}. {row[0]}")
    print(f"Link: {row[1]}\n")

table = pd.DataFrame(records, columns=["Title", "Website"])
table.to_csv("aaft_search_results.csv", index=False)

print("CSV file saved successfully")

browser.save_screenshot("aaft_bing_page.png")
print("Screenshot captured")

browser.quit()
```

```python
Import required libraries

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

Start the browser

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
# Opens Chrome browser and automatically installs matching driver

Open a website

driver.get("https://www.bing.com/")
# Opens Bing website

Browser navigation

driver.back()
# Goes to previous page

driver.forward()
# Goes to next page

driver.refresh()
# Reloads the current page

Close browser

driver.close()
# Closes current tab

driver.quit()
# Closes complete browser session

Find elements on webpage

from selenium.webdriver.common.by import By

driver.find_element(By.ID, "id")
# Find using ID

driver.find_element(By.NAME, "name")
# Find using name attribute

driver.find_element(By.CLASS_NAME, "class")
# Find using class name

driver.find_element(By.TAG_NAME, "tag")
# Find using tag name

driver.find_element(By.LINK_TEXT, "text")
# Find link using exact text

driver.find_element(By.PARTIAL_LINK_TEXT, "text")
# Find link using partial text

driver.find_element(By.CSS_SELECTOR, "css")
# Find using CSS selector

driver.find_element(By.XPATH, "xpath")
# Find using XPath

driver.find_elements(By.CLASS_NAME, "item")
# Finds multiple matching elements

Element actions

element = driver.find_element(By.ID, "id")

element.click()
# Clicks the element

element.send_keys("Hi")
# Types text inside input field

element.clear()
# Clears existing text

element.submit()
# Submits form

Keyboard actions

from selenium.webdriver.common.keys import Keys

element.send_keys(Keys.RETURN)
# Press Enter key

element.send_keys(Keys.CONTROL, 'a')
# Select all text

Waits in Selenium

driver.implicitly_wait(10)
# Waits up to 10 seconds for elements to load

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

WebDriverWait(driver, 10).until(
    EC.presence_of_element_located((By.ID, "id"))
)
# Waits until element appears on screen

Get information from webpage

element.text
# Gets visible text

element.get_attribute("href")
# Gets attribute value

driver.title
# Gets page title

driver.current_url
# Gets current page URL

driver.page_source
# Gets complete HTML source

Working with windows and tabs

driver.current_window_handle
# Current window ID

driver.window_handles
# List of all opened windows

driver.switch_to.window(driver.window_handles[0])
# Switch to another window

driver.switch_to.new_window('tab')
# Opens new tab

Scrolling the webpage

driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
# Scroll to bottom of page

Run JavaScript

driver.execute_script("alert('Hello')")
# Runs JavaScript code

Frames and iFrames

driver.switch_to.frame("frame_name")
# Switch inside frame

driver.switch_to.default_content()
# Return to main page

Cookies

driver.get_cookies()
# Get all cookies

driver.add_cookie({"name": "test", "value": "123"})
# Add new cookie

driver.delete_all_cookies()
# Delete all cookies

Take screenshot

driver.save_screenshot("NameForImage.png")
# Saves screenshot

Mouse actions

from selenium.webdriver.common.action_chains import ActionChains

actions = ActionChains(driver)

actions.move_to_element(element).click().perform()
# Move mouse and click

Dropdown handling

from selenium.webdriver.support.ui import Select

select = Select(element)

select.select_by_visible_text("Option")
# Select using visible text

select.select_by_index(1)
# Select using index

select.select_by_value("value")
# Select using value

Handle alerts

alert = driver.switch_to.alert

alert.accept()
# Click OK button

alert.dismiss()
# Click Cancel button

alert.text
# Gets alert message

Page timeout

driver.set_page_load_timeout(10)
# Maximum wait time for page loading

Check element state

element.is_displayed()
# Checks if element is visible

element.is_enabled()
# Checks if element is enabled

element.is_selected()
# Checks if checkbox/radio is selected

JavaScript click

driver.execute_script("arguments[0].click();", element)
# Force click using JavaScript
```
