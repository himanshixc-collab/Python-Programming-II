```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def display_info(self):
        print(f"{self.year} {self.brand} {self.model}")


class ElectricCar(Car):
    def __init__(self, brand, model, year, battery_size):
        super().__init__(brand, model, year)
        self.battery_size = battery_size

    def display_battery(self):
        print(f"brand: {self.brand} year: {self.year} model:{self.model} Battery Size: {self.battery_size} kWh")

my_car = ElectricCar("Tesla", "Model 3", 2023, 75)
my_car.display_battery()
```

```python
class Computer:
    def __init__(self, brand, processor, ram):
        self.brand = brand
        self.processor = processor
        self.ram = ram


class Laptop(Computer):
    def __init__(self, brand, processor, ram, battery_life):
        super().__init__(brand, processor, ram)
        self.battery_life = battery_life

    def display_specs(self):
        print(f"{self.brand}, {self.processor}, {self.ram}GB RAM, Battery Life: {self.battery_life} hours")
        

my_laptop = Laptop("Dell", "Intel i7", 16, 10)
my_laptop.display_specs()
```

```python
class ElectronicDevice:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price


class Smartphone(ElectronicDevice):
    def __init__(self, brand, price, operating_system):
        super().__init__(brand, price)
        self.operating_system = operating_system

    def display_info(self):
        print(f"{self.brand} Smartphone running on {self.operating_system} and it's price is {self.price}")

my_phone = Smartphone("Samsung", 999, "Android")
my_phone.display_info()

class Tablet(ElectronicDevice):
    def __init__(self, brand, price, operating_system):
        super().__init__(brand, price)
        self.operating_system = operating_system

    def display_info(self):
        print(f"{self.brand} Tablet running on {self.operating_system} and it's price is {self.price}")

my_tablet = Tablet("Samsung", 999, "Android")
my_tablet.display_info()
```

```python
class Artist:
    def __init__(self, name, field):
        self.name = name
        self.field = field


class Musician(Artist):
    def __init__(self, name, field, instrument_played):
        super().__init__(name, field)
        self.instrument_played = instrument_played

    def display_info(self):
        print(f"{self.name} is a {self.field} who plays {self.instrument_played}")


class Painter(Artist):
    def __init__(self, name, field, instrument_played):
        super().__init__(name, field)
        self.instrument_played = instrument_played

    def display_info(self):
        print(f"{self.name} is a {self.field} known for using {self.instrument_played}")


musician1 = Musician("John Mayer", "musician", "guitar")
musician1.display_info()

painter1 = Painter("Leonardo da Vinci", "painter", "oil paints")
painter1.display_info()
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author


class FictionBook(Book):
    def __init__(self, title, author, genre):
        super().__init__(title, author)
        self.genre = genre

    def display_info(self):
        print(f"'{self.title}' by {self.author} - Genre: {self.genre}")


class NonFictionBook(Book):
    def __init__(self, title, author, genre):
        super().__init__(title, author)
        self.genre = genre

    def display_info(self):
        print(f"'{self.title}' by {self.author} - Genre: {self.genre}")

fiction1 = FictionBook("Harry Potter and the Sorcerer's Stone", "J.K. Rowling", "Fantasy")

nonfiction1 = NonFictionBook("Sapiens: A Brief History of Humankind", "Yuval Noah Harari", "History")
fiction1.display_info()
nonfiction1.display_info()
```

```python
class Student:
    def __init__(self, name, marks1, marks2, marks3):
        self.name = name
        self.marks1 = marks1
        self.marks2 = marks2
        self.marks3 = marks3

    def calculate_average(self):
        average = (self.marks1 + self.marks2 + self.marks3) / 3
        print(f"{self.name}'s Average Marks: {average}")

student1 = Student("Alice", 85, 90, 78)
student1.calculate_average()
```
