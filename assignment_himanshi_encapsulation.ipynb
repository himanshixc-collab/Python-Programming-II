```python
class Laptop:
    def __init__(self, price):
        self.__price = price

    def old_price(self):
        return self.__price

    def set_price(self, new_price):
        if new_price > 0:
            self.__price = new_price
        else:
            print("Price must be positive.")

    def get_price(self):
        return self.__price

laptop = Laptop(1000)
print("old price was", laptop.get_price())  # Get the current price

laptop.set_price(1200)  # Update the price
print("new price is", laptop.get_price())  # Get the updated price
```

```python
class BankAccount:
    def __init__(self, initial_balance=0):
        self.__balance = initial_balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            print("Invalid withdrawal amount.")

    def balance(self):
        return self.__balance

account = BankAccount(500)
account.deposit(200)
account.withdraw(100)
print("balance is",account.balance())
```

```python
class Movie:
    def __init__(self, rating=1):
        self.__rating = 1
        self.set_rating(rating)

    def set_rating(self, rating):
        if 1 <= rating <= 10:
            self.__rating = rating
        else:
            print("Rating must be between 1 and 10.")

    def get_rating(self):
        return self.__rating

movie = Movie(8)
movie.set_rating(9)
print("the movie has ratings of",movie.get_rating())
```

```python
class Book:
    def __init__(self, title, author):
        self.__title = title
        self.__author = author

    def book_title(self):
        return self.__title

    def book_author(self):
        return self.__author

book = Book("1984", "George Orwell")
print("the title of the book is" ,book.book_title() ,"and the author is" , book.book_author())
```
