```python
class RacingCar:
    def _init_(self):
        self.speed = 0  

 
    def acceleration(self, total_time):
        if total_time > 0:
            acceleration = self.speed / total_time
            print("Acceleration:", acceleration, "m/s²")
        else:
            print("Time must be greater than 0")

 
    def braking(self, brake_value):
        self.speed = self.speed - brake_value
        if self.speed < 0:
            self.speed = 0
        print("Speed after braking:", self.speed)

car = RacingCar()

car.speed = 100   
car.acceleration(10)  
car.braking(30)
```

```python
class Diet:
    def __init__(self):   # ✅ correct constructor
        self.calories = 0

    def junk_calories(self, junk):
        self.calories = self.calories + junk

    def exercise(self, burn):
        self.calories = self.calories - burn
        if self.calories < 0:
            self.calories = 0

    def check_health(self):
        if self.calories > 30:
            print("Unhealthy")
        else:
            print("Healthy")


d = Diet()
d.junk_calories(40)
d.check_health()   # Unhealthy
d.exercise(20)
d.check_health()   # Healthy
```

```python
class IMBD:
    def _init_(self):
        self.rating = 0

    def new_rating(self, new_rating):
        self.rating = new_rating


movie = IMBD()
movie.new_rating(8.5)
print(movie.rating)
```

```python
class BankAccount:
    def __init__(self, balance):   
        self.__balance = balance

    # getting balance from a method
    def get_balance(self):
        return self.__balance

    # deposit rule
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Amount deposited")
        else:
            print("Invalid deposit")

    # withdraw rule
    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawal successful")
        else:
            print("Insufficient balance")


acc = BankAccount(1000)
acc.deposit(500)        
acc.withdraw(200)       
print(acc.get_balance())
```
