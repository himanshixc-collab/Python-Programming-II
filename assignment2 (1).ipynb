GENERATOR

```python
def my_colors():
    yield "red"
    yield "blue"
    yield "green"

gen = my_colors()
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
def count_up(n):
    for i in range(1, n+1):
        yield i

gen = count_up(10)
for num in gen:
    print(num)
```

```python
gen_exp = (x**3 for x in range(11))
print(next(gen_exp))
print(next(gen_exp))
print(next(gen_exp))
print(next(gen_exp))
print(next(gen_exp))
print(next(gen_exp))
print(next(gen_exp))
print(next(gen_exp))
```

```python
def numbers():
    yield 1
    yield 2
    yield 3

Nums = numbers()
for num in Nums:
    print(num)
```

```python
def counter(n):
    while n > 0:
        val = (yield n)
        if val is not None:
            n = val
        else:
            n -= 1

gen = counter(5)
print(next(gen))
print(gen.send(10))
print(next(gen))
```

DECORATOR

```python
def decor(fun):
  def ginger():

    print("hello,i am")
    fun()

  return ginger

@decor
def ben():
    print("Ben")

ben()
```

```python
def decor(fun):
  def ginger():

    print("start")
    fun()
    print("success")

  return ginger

@decor
def progg():
  print("loadin")
  print("almost done")

progg()
```

```python
def decor(func):
    def ginger(a, b):
        print(f"Calling {func.__name__} with arguments: {a}, {b}")
        return func(a, b)
    return ginger

@decor
def addition(x, y):
    return x + y

print(addition(22, 5))
```

```python
def decor(func):
    def ginger(user):
        if user != "teacher":
            print("Access denied")
            return
        return func(user)
    return ginger

@decor
def access_dashboard(user):
    print(f"{user} accessed the dashboard.")

access_dashboard("student")
access_dashboard("teacher")
```

```python
def decor(func):
  def wrapper(*args, **kwargs):
    result = func(*args, **kwargs)
    return result*2
  return wrapper

@decor
def get_number():
  return 10

print(get_number())
```
